## Data preparation

```python
import pandas as pd
```

```python
def get_data_worldbank_xls(path):

    df = pd.read_excel(
        pd.ExcelFile(path),
        sheet_name='Data',
        skiprows=3
    )

    df.drop(columns=["Country Name", "Indicator Name", "Indicator Code", "1960", "1961", "1962", "1963", "1964", "1965", "1966", "1967", "1968", "1969", "1970", "1971", "1972", "1973", "1974", "1975", "1976", "1977", "1978", "1979", "1980", "1981", "1982", "1983", "1984", "1985", "1986", "1987", "1988", "1989", "1990", "1991", "1992", "1993", "1994", "1995", "1996", "1997", "1998", "1999", "2025"], inplace=True)

    df.dropna(
        thresh=len(df.columns) // 2,
        inplace=True
    )

    return df
```

```python
df_ed_gov_exp = get_data_worldbank_xls('data/education_gov_expenditure_%.xls')

df_ed_post_sec = get_data_worldbank_xls('data/education_post_secondary_%.xls')

df_ec_gni_pc = get_data_worldbank_xls('data/economic_gni_per_capita.xls')

df_ec_gini_idx = get_data_worldbank_xls('data/economic_gini_index.xls')

df_urban_pop = get_data_worldbank_xls('data/urban_population_%.xls')
```
